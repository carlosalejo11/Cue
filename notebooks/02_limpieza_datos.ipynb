```python
import numpy as np
import pandas as pd
print("Librerías cargadas correctamente")
```

```python
df = pd.read_csv('../data/raw/tmdb_5000_movies.csv')
print(f"Dataset cargado: {df.shape[0]} filas, {df.shape[1]} columnas")
```

```python
cols_to_drop = ['homepage', 'tagline', 'overview', 'spoken_languages', 'status']
df = df.drop(columns = cols_to_drop)
print(f"Columnas restantes: {df.shape[1]}")
print(df.columns.tolist())
```

```python
df = df.dropna(subset=['release_date', 'runtime'])
print(f"Filas restantes despues de eliminar valores nulos: {df.shape[0]}")
```

```python
df = df[(df['budget'] > 0) & (df['revenue'] > 0)]
print(f"Filas restantes despues de filtrar ceros: {df.shape[0]}")
```

```python
df['release_date'] = pd.to_datetime(df['release_date'])
df['release_year'] = df['release_date'].dt.year
print(f"Tipo de dato de release_date: {df['release_date'].dtype}")
print(df[['title', 'release_date', 'release_year']].head())
```

```python
df['runtime'] = df['runtime'].astype(int)
print(f"El tipo de dato runtime es: {df['runtime'].dtype}")
print(df['runtime'].describe())
```

```python
df['net_profit'] = df['revenue'] - df['budget']
df['roi'] = ((df['revenue'] - df['budget']) / df['budget'] * 100).round(2)
print(df[['title', 'budget', 'revenue', 'net_profit', 'roi']].head())
```

```python
df.to_csv('../data/processed/tmdb_clean.csv', index = False)
print("El dataset ha sido exportado correctamente")
print(f"Dataset Shape: {df.shape}")
```
