```python
import pandas as pd
import numpy as np
print("Librerías cargadas")
```

```python
df_cameras = pd.read_csv('../data/raw/cinema_cameras.csv')
print(f"Shape: {df_cameras.shape}")
print(f"Columnas: ")
print(df_cameras.columns.tolist())
```

```python
print(df_cameras.info())
```

```python
print(df_cameras.isnull().sum())
```

```python
print(df_cameras['category'].value_counts())
print("--------")
print(df_cameras['brand'].value_counts())
```

```python
print(df_cameras['category'].unique())
print("-------")
print(df_cameras[df_cameras['brand'].str.contains('Canon', na=False)]['brand'].unique())
```

```python
# Corrección de espacios y categorías para unificar
df_cameras['brand'] = df_cameras['brand'].str.strip()
df_cameras['category'] = df_cameras['category'].str.strip()

df_cameras['category'] = df_cameras['category'].replace({'Cinema': 'Digital Cinema', 'Mirrorless Cine': 'Mirrorless'})

print(df_cameras['category'].value_counts())
print("-----")
print(df_cameras['brand'].value_counts())
```

```python
df_cameras.to_csv('../data/processed/cinema_cameras_clean.csv', index=False)
print("Dataset exportado exitosamente")
print(f"Shape: {df_cameras.shape}")
```
