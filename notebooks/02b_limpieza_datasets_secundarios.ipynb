```python
# Exploración de dataset IMDB movies
df_imdb = pd.read_csv('../data/raw/IMDb movies.csv')
print(f"Dataset Shape: {df_imdb.shape}")
print(f"\nColumnas:")
print(df_imdb.columns.tolist())
```

```python
# Revision de null y dtype
print(df_imdb.isnull().sum()) 
print(f"\nTipos de datos:")
print(df_imdb.dtypes)
```

```python
# Dataset IMDb names
df_names = pd.read_csv('../data/raw/IMDb names.csv')
print(f"Dataset Shape: {df_names.shape}")
print(f"\nColumnas")
print(df_names.columns.tolist())
print(f"\nDataset Head:")
df_names.head()
```

```python
print(df_imdb.equals(df_names))
```

```python
# Comprobar datos de movies y names
# Se descartara names
print(f"IMDb movies Shape: {df_imdb.shape}")
print(f"IMDb names Shape: {df_names.shape}")
print(f"\nFila 1 IMDb movies:")
print(df_imdb.iloc[0]['title'])
print(f"\nFila 1 IMDb names:")
print(df_names.iloc[0]['title'])
```
